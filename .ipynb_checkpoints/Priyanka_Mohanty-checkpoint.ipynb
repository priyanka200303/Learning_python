## Q-1 What is the difference between append() and extend() in lists?

```python
list1=[1,2,3]
list1.append([4,5])
print("Using append():", list1)
list1.extend([6,7])
print("Using extend():", list1)
```

### append() adds the given element as a single item at the end of the list. It doesnot matter whether the element is a number, string or another list. it is added as one complete unit. If we append a list it becomes a nested list inside the original list. On other hand extend() is used when we want to add multilpe elements individualy to a list. If we extend a list then it extend a list by inserting elements one by one instead of creating a nested list. For example append([4,5]) creates a nested list but extend([6,7]) adds 6, 7 separately.
### Time complexity:append()=O(1) and expand()=O(n).

## Q-2 Write a program to merge two dictionaries.(take any examples of your wish)

```python
dict1={"a": 1,"b": 2}
dict2={"b": 3,"c": 4}
merged_dict={**dict1, **dict2}
print("Merged Dictionary:",merged_dict)
# Another method
dict1.update(dict2)
print(dict1)
```

### We can merge two dictionary by simply combining their key-values pairs using methods like {**dict1, **dict2} or update(). If both dictionaries have the same key then the value from the second dictionary will replace the first.For example in dict1 we have b=2 and in dict2 we have b=3 and after merging two dictionary we got the key b=3. 
### Time complexity:O(n)

## Q-3 Write a function that returns indices of two numbers whose sum equals the target.
## Input>nums=[4,7,1,8,5]
## Output>target=9

```python
nums=[4, 7, 1, 8, 5]
target=9
for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i] + nums[j] == target: 
                print("Indices:", i,j)

```

### We solve this by checking every possible pair of numbers using two loops. For each pair, we add them and see if the sum equals the target. Whenever the condition is satisfied, we print their indices. This way, we ensure we don't miss any valid pair.
### Time complexity:O(n^2)

## Q-4 Check if a string of brackets is valid.
## Input>s="{[()]}"
## Output>True

```python
s="{[()]}"
stack=[]
pairs={')':'(',']':'[','}':'{'}
for ch in s:
    if ch in "{[(":
        stack.append(ch)
    else:
        if stack and stack[-1]==pairs[ch]:
            stack.pop()
        else:
            print(False)
            break
else:
    print(len(stack)==0)
        
```

### We use a stack to keep track of opening brackets. Whenever we get a closing bracket, we check if it matches the last opened one. If everything matches properly and nothing is left unmatched, the string is valid. 
### Time complexity:O(n)

## Q-5 Find the first negative number in every window of size k.
## Input>arr=[12,-1,-7,8,-15,30,16,28] and k=3
## Expected Output>[-1,-1,-7,-15,-15,0]


```python
arr=[12,-1,-7,8,-15,30,16,28]
k=3
n=len(arr)
for i in range(n-k+1):
    found=False

    for j in range(i,i+k):
        if arr[j]<0:
            print(arr[j], end=" ")
            found=True
            break
    if not found:
            print(0,end=" ")
```

### We take each window of size k and check its elements one by one. As soon as we find the first negative number we print it, if none is found we print 0.
### O(nxk)

## Q-6 Determine whether a linked list contains a cycle & what's multi threading and multi-processing?

```python
class Node:
    def __init__(self, data):
        self.data=data
        self.next=None
def has_cycle(head):
    slow=head
    fast=head
    while fast and fast.next:
        slow=slow.next
        fast=fast.next.next
        if slow==fast:
            return True
    return False
```

### We use two pointers—one moves one step at a time and the other moves two steps. If the linked list has a cycle, both pointers will eventually meet; otherwise, the fast pointer will reach the end.
### Multithreading vs Multiprocessing: Multithreading means running multiple threads within the same process to perform tasks concurrently, mainly useful for I/O tasks. Multiprocessing means running multiple processes independently, which is useful for CPU-intensive tasks and allows true parallel execution.
### Time complexity:O(n)

## Q-7 Py code to find the Greatest Common Divisor(GCD) of two given numbers 36 and 60. 

```python
a=36
b=60
while b!=0:
    temp=b
    b=a%b
    a=temp
print("GCD of 36 and 60 is:", a)
```

### We repeatedly replace the larger number with the reminder until one number becomes zero. The remaining number is GCD.
### Time complexity:O(log n)

## Q-8 Find the first occurrence of an element in a sorted array
## Input>arr=[1,2,2,2,3,4]
## Output>1

```python
arr=[1,2,2,2,3,4]
target=2
for i in range(len(arr)):
    if arr[i] == target:
        print("Index:", i)
        break
```

### We go through the array from the beginning and stop as soon as we find the target element. Since the array is sorted, the first match we find is the first occurance.
### Time complexity:O(n)

## Q-9 Write functions for:
## i.Inorder transversal
## ii.Preorder transversal
## iii.Postorder transversal

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Inorder Traversal (Left, Root, Right)
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)


# Preorder Traversal (Root, Left, Right)
def preorder(root):
    if root:
        print(root.data, end=" ")
        preorder(root.left)
        preorder(root.right)


# Postorder Traversal (Left, Right, Root)
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data, end=" ")


# Example Tree
root = Node(1)
root.left = Node(2)
root.right = Node(3)

print("Inorder:")
inorder(root)

print("\nPreorder:")
preorder(root)

print("\nPostorder:")
postorder(root)
```

### In inorder traversal, we first visit the left subtree, then the root node, and finally the right subtree, which is useful for getting sorted output in a binary search tree. In preorder traversal, we visit the root first, then move to the left and right subtrees, which is helpful for copying or creating a tree. In postorder traversal, we visit the left and right subtrees before the root, which is useful when deleting or evaluating a tree.
### Time complexity:O(n)
