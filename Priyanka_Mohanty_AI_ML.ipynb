### In this question, I will perform exploratory data analysis to understand the key factors affecting house prices.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### I imported necessary libraries for data handling, visualizations, and analysis.

```python
df = pd.read_csv(r"C:\Users\priya\Desktop\Internship\Datasets\kc_house_data.csv")
df.head()
```

### I checked data types, number of entries, and summary statistics to understand the distribution of variables.

```python
df = df.drop(['id', 'date', 'zipcode'], axis=1) 
```

### The id column is just an identifier and does not help in analysis. The date column is not directly relevent for price prediction in this task. The zipcode is representing the location but not useful for correlation analysis. so these three were removed.

```python
df.isnull().sum()
```

### I checked for missing values in the dataset. There were only one column sqft_above has 2 missing values and no significant missing values in other columns.Since the missing value numbers is very small so i will fill it with median. This will maintain the data consistency without losing informations.

```python
df['sqft_above']=df['sqft_above'].fillna(df['sqft_above'].median())
```

### I filled the 2 missing values by median.

```python
sns.boxplot(x=df['price'])
plt.show()
```

### The boxplot shows a right skewed distribution with many high value outliers. That indicates the presence of extreme house prices.

```python
df=df[df['price'] < df['price'].quantile(0.99)]
```

### I removed the top 1% extreme price values to make the analysis more realistic.

```python
sns.boxplot(x=df['price'])
plt.show()
```

### This is the plot after removing the outliers.

```python
corr=df.corr(numeric_only=True)
plt.figure()
sns.heatmap(corr, cmap='coolwarm')
plt.show()
```

### I used a heatmap to understand the relationships between different variables and house price.

```python
corr['price'].sort_values(ascending=False)
```

### The correlation results shows that the features like grade, sqft_living and sqft_above have the strongest relationship with price while others have weaker affect. This means house quality and living area size are the main factors that affects the price. 





# Q-3 Time Series Forcasting - AirPassengers Dataset. 

### In this task, I performed time series forcasting to predict future passenger values using ARIMA model.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\priya\Desktop\Internship\Datasets\AirPassengers.csv")
df.head()
```

### I loaded the dataset and displayed few rows to understand the data.

```python
df.columns
```

```python
df['Month'] = pd.to_datetime(df['Month'])
df.set_index('Month', inplace=True)
```

### I converted the Month column into datetime format and set it as index.

```python
plt.figure()
plt.plot(df)
plt.title("Air Passengers over time")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()
```

### The plot shows a clear upward trend and repeating seasonal pattern in passenger data.

```python
train_size = int(len(df) * 0.8)

train = df[:train_size]
test = df[train_size:]

print("Train size:", len(train))
print("Test size:", len(test))
```

### I divided the dataset into training and testing sets to evaluate model performance.

```python
from statsmodels.tsa.arima.model import ARIMA
model= ARIMA(train, order=(1,1,1))
model_fit= model.fit()
```

### I used the ARIMA model to capture trend and patterns in the time series data.

```python
forecast=model_fit.forecast(steps=len(test))
```

```python
import numpy as np
from sklearn.metrics import mean_squared_error
rmse=np.sqrt(mean_squared_error(test, forecast))
print("RMSE:", rmse)
```

### The RMSE value is 97.49, which means the predicted values are on average about 97 passengers away from the actual values.Since the passenger numbers are in the hundreds, this error is not very large.It shows that the model is understanding the pattern, but there is still some difference between prediction and actual data.
